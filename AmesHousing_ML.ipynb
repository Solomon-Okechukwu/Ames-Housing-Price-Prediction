```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer
from sklearn.linear_model import LinearRegression
```

```python
numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical:", len(numeric_features))
print("Categorical:", len(categorical_features))
```

```python
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("to_string", FunctionTransformer(lambda x: x.astype(str))),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])
```

```python
preprocessor = ColumnTransformer([
    ("categorical", categorical_pipeline, categorical_features),
    ("numerical", numeric_pipeline, numeric_features)
])
```

```python
model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])
```

```python
model.fit(X_train, y_train)
```

```python
#NOTE#

╔══════════════════════════════════════════════════════════════╗
║        MACHINE LEARNING MODEL EVALUATION — QUICK NOTE        ║
║              Ames Housing Price Prediction                   ║
╚══════════════════════════════════════════════════════════════╝

MODEL:
Linear Regression

TARGET:
SalePrice — the house price we want the model to predict.

PREDICTION:
y_pred = model.predict(X_test)

The model uses the house characteristics in X_test to predict
the SalePrice of houses it has not seen during training.


──────────────────────────────────────────────────────────────
1. MAE — Mean Absolute Error
──────────────────────────────────────────────────────────────

MAE tells us the average amount by which the model's predictions
are different from the actual house prices.

Our result:
MAE = 17,849.15

Meaning:
The model's predictions are approximately $17,849 away from the
actual house prices on average.

LOWER MAE = BETTER


──────────────────────────────────────────────────────────────
2. MSE — Mean Squared Error
──────────────────────────────────────────────────────────────

MSE calculates the average squared difference between predicted
prices and actual prices.

Our result:
MSE = 922,770,239.35

Because the errors are squared, the number becomes very large.
Therefore, MSE is usually less intuitive to interpret directly.

LOWER MSE = BETTER


──────────────────────────────────────────────────────────────
3. RMSE — Root Mean Squared Error
──────────────────────────────────────────────────────────────

RMSE is the square root of MSE.

It measures prediction error in the SAME UNIT as SalePrice,
making it easier to understand than MSE.

Our result:
RMSE = 30,377.13

Meaning:
The model's typical prediction error is around $30,377, with
larger errors receiving more importance.

LOWER RMSE = BETTER


──────────────────────────────────────────────────────────────
4. R² — R-Squared
──────────────────────────────────────────────────────────────

R² measures how much of the variation in house prices is
explained by the model.

Our result:
R² = 0.8849

This is approximately:

88.49%

Meaning:
The model explains about 88.5% of the variation in house
prices in the TEST dataset.

Generally:
R² closer to 1 = better
R² closer to 0 = weaker

An R² of 0.8849 is a strong result for this Linear Regression
model.


──────────────────────────────────────────────────────────────
5. ACTUAL vs PREDICTED
──────────────────────────────────────────────────────────────

ACTUAL PRICE:
The real SalePrice from the dataset.

PREDICTED PRICE:
The price estimated by our machine learning model.

Example:

Actual:     $161,000
Predicted:  $142,230

Difference:
Approximately $18,770


──────────────────────────────────────────────────────────────
6. OUR MODEL'S RESULTS
──────────────────────────────────────────────────────────────

MAE  = $17,849.15
RMSE = $30,377.13
R²   = 0.8849 (88.49%)

Overall interpretation:

The Linear Regression model is performing reasonably well.
It explains approximately 88.5% of the variation in house
prices, while its average absolute prediction error is about
$17,849.

However, the model can potentially be improved by trying
different algorithms and better feature engineering.


──────────────────────────────────────────────────────────────
7. QUICK RULE TO REMEMBER
──────────────────────────────────────────────────────────────

MAE  → Average prediction error
MSE  → Squared prediction error
RMSE → Error measure in the original price unit
R²   → How much variation the model explains

For MAE, MSE and RMSE:
LOWER = BETTER

For R²:
HIGHER = GENERALLY BETTER


──────────────────────────────────────────────────────────────
8. OUR CURRENT ML WORKFLOW
──────────────────────────────────────────────────────────────

Load data
    ↓
Understand features
    ↓
Clean data
    ↓
Separate X and y
    ↓
Split into training and testing data
    ↓
Handle missing values
    ↓
Encode categorical variables
    ↓
Train Linear Regression
    ↓
Make predictions
    ↓
Evaluate MAE, MSE, RMSE and R²
    ↓
Improve the model
    ↓
Try other algorithms
    ↓
Compare models

Your key result to remember: R² = 0.8849, meaning your current Linear Regression model explains about 88.5% of the variation in house prices on the test set.
```
